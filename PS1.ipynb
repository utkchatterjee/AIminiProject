# **Taking Input**

```python
R,C=map(int,input().split())
grid=[['S' for c in range(C)] for r in range(R)]
for r in range(R):
  grid[r]=list(input())
print(f"{grid}")
```

```python
#Finding the start node
for i in range(R):
    for j in range(C):
      if(grid[i][j]=='S'):
        startRow,startCol=i,j
```

# *DFS*

```python
def DFS(row,column,path,visited):
  visited.append((row,column))
  if(grid[row][column] == 'G'):
   return True

  if((row<R-1 and (row+1,column) not in visited) and grid[row+1][column] != '#'):
    if(DFS(row+1,column,path,visited)):
      path.append("DOWN")
      return True
  if((row>0 and (row-1,column) not in visited)and grid[row-1][column] != '#'):
    if(DFS(row-1,column,path,visited)):
      path.append("UP")
      return True
  if((column<C-1 and (row,column+1) not in visited) and grid[row][column+1] != '#'):
    if(DFS(row,column+1,path,visited)):
      path.append("RIGHT")
      return True
  if((column>0 and (row,column-1) not in visited) and grid[row][column-1] != '#'):
    if(DFS(row,column-1,path,visited)):
      path.append("LEFT")
      return True
  return False
```

```python
path=[]
visited=[]
if(DFS(startRow,startCol,path,visited)):
  path.reverse()
  print(f"For DFS\nPath has been found.\nPath is {path}\nNumber of moves={len(path)}\nNodes expanded={len(visited)}\n")
else:
  print("No path found\n")
```

# **BFS**

```python
#BFS alter
from collections import deque
def BFS(row,col,came_from):
  queue=deque()
  queue.append((row,col))
  visited=[]
  while(queue):
    row,col=queue.popleft()
    if (grid[row][col]=='G'):
      return row,col
    #append all unvisited neighbors
    neighbors=[(row+1,col),(row-1,col),(row,col+1),(row,col-1)]
    for (neighborRow,neighborCol) in neighbors:
      if ((neighborRow,neighborCol) not in visited and 0<=neighborRow<R and 0<=neighborCol<C and grid[neighborRow][neighborCol]!='#'):
        queue.append((neighborRow,neighborCol))
        came_from[(neighborRow,neighborCol)]=(row,col)
    #this node is now visited, let's get the next one
    visited.append((row,col))

  return R+1,C+1

def pathBFS():
  came_from={} #child->parent
  (row,col)=BFS(startRow,startCol,came_from)
  if(row==R+1):
    return []
  path=[]
  #print(f"Is this even extant? {came_from}")
  while((row,col)!=(startRow,startCol)):
    path.append((row,col))
    row,col=came_from[(row,col)]
  path.append((startRow,startCol))
  return path
```

```python
#Print Alter
path=pathBFS()
if len(path) == 0:
  print("No path found\n")
else:
  print(f"For BFS\nPath has been found.\nPath is");
  pathInWords=[]
  prevRow=startRow
  prevCol=startCol
  path.reverse()
  for element in path:
    row,column=element
    if(row>prevRow):
      pathInWords.append("DOWN")
    elif(row<prevRow):
      pathInWords.append("UP")
    elif(column>prevCol):
      pathInWords.append("RIGHT")
    elif(column<prevCol):
      pathInWords.append("LEFT")
    prevRow=row
    prevCol=column
  print(f"{pathInWords}\nNumber of moves={len(pathInWords)}\nNodes expanded={len(visited)}\n")

```
