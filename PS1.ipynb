# **Taking Input**

```python
R,C=map(int,input().split())
grid=[['S' for c in range(C)] for r in range(R)]
for r in range(R):
  grid[r]=list(input())
print(f"{grid}")
```

# *DFS*

```python
path=[]
visited=[]
def DFS(row,column,path=path,visited=visited):
  visited.append((row,column))
  if(grid[row][column] == 'G'):
   return True

  if((row<R-1 and (row+1,column) not in visited) and grid[row+1][column] != '#'):
    if(DFS(row+1,column)):
      path.append("RIGHT")
      return True
  if((row>0 and (row-1,column) not in visited)and grid[row-1][column] != '#'):
    if(DFS(row-1,column)):
      path.append("LEFT")
      return True
  if((column<C-1 and (row,column+1) not in visited) and grid[row][column+1] != '#'):
    if(DFS(row,column+1)):
      path.append("DOWN")
      return True
  if((column>0 and (row,column-1) not in visited) and grid[row][column-1] != '#'):
    if(DFS(row,column-1)):
      path.append("UP")
      return True
  return False
```

```python
path=[]
visited=[]
if(DFS(0,0,path,visited)):
  path.reverse()
  print(f"For DFS\nPath has been found.\nPath is {path}\nNumber of moves={len(path)}\nNodes expanded={len(visited)}\n")
else:
  print("No path found\n")
```

# **BFS**

```python
path=[]
from collections import deque
sequence=deque()
def BFS(row,column):
  if(grid[row][column] == 'G'):
    return True
  visited.append((row,column))
  if((row<R-1 and (row+1,column) not in visited) and grid[row+1][column] != '#'):
    sequence.append((row+1,column))
  if((row>0 and (row-1,column) not in visited)and grid[row-1][column] != '#'):
    sequence.append((row-1,column))
  if((column<C-1 and (row,column+1) not in visited) and grid[row][column+1] != '#'):
    sequence.append((row,column+1))
  if((column>0 and (row,column-1) not in visited) and grid[row][column-1] != '#'):
    sequence.append((row,column-1))
  if(len(sequence)==0):
    return False
  row,column=sequence.popleft()
  if(BFS(row,column)):
    path.append((row,column))
    return True
```

```python
visited=[]
if(BFS(0,0)):
  print(f"For BFS\nPath has been found.\nPath is");
  pathInWords=[]
  prevRow=0
  prevCol=0
  path.reverse()
  for element in path:
    row,column=element
    if(row>prevRow):
      pathInWords.append("DOWN")
    elif(row<prevRow):
      pathInWords.append("UP")
    elif(column>prevCol):
      pathInWords.append("RIGHT")
    else:
      pathInWords.append("LEFT")
    prevRow=row
    prevCol=column
  print(f"{pathInWords}\nNumber of moves={len(pathInWords)}\nNodes expanded={len(visited)}\n")
else:
  print("No path found\n")
```
